```python
while True:
    number = input("Enter your phone number: ")

    # Checks if the number follows the correct format
    if (len(number) == 13 and
        number[0] == "0" and
        number[1] == "9" and
        number[4] == "-" and
        number[8] == "-" and
        number[:4].isdigit() and # Checks if all the Xs are numbers
        number[5:8].isdigit() and
        number[9:].isdigit()):

        print("Phone number accepted:", number)
        break
    else:
        print("Invalid format! Please enter your phone number in 09XX-XXX-XXXX format.")
```
