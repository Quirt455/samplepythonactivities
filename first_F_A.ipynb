```python

import math

x1 = float(input("Enter x1"))
y1 = float(input("Enter y1"))
x2 = float(input("Enter x2"))
y2 = float(input("Enter y2"))

diff_x = x1 - x2
diff_y = y1 - y2

squared_diff_x = math.pow(diff_x,2)
squared_diff_y = math.pow(diff_y,2)

sum_of_square_diffs = squared_diff_x + squared_diff_y

distance = math.sqrt(sum_of_square_diffs)

print(distance)


#Using a library like 'math' is practical because it gives pre-built, easy to use, and tested functions,
# saving some time from coding and reducing the errors. For example, 'math.sqrt()' and 'math.pow()' allowed me to
# calculate square roots and powers without having to use really long mathematical algorithms.





```
